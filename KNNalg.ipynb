```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
# Generate synthetic dataset
n_classes = 6
data, labels = make_blobs(n_samples=1000,
centers=n_classes,
cluster_std=1.3,
random_state=1)
# Define colors for each class
colours = ('green', 'red', 'blue', 'magenta', 'yellow', 'pink')
# Plot the dataset
fig, ax = plt.subplots()
for n_class in range(n_classes):
  ax.scatter(data[labels == n_class, 0], data[labels == n_class, 1],
c=colours[n_class], s=10, label=str(n_class))
ax.set_title('Generated Data Points')
ax.legend()
plt.show()
# Split the data into training and testing sets
res = train_test_split(data, labels,
train_size=0.7,
test_size=0.3,
random_state=1)
train_data, test_data, train_labels, test_labels = res
print(len(train_data), len(test_data), len(train_labels))
# Initialize lists to store accuracy scores
X, Y = [], []
for k in range(1, 25):
  classifier = KNeighborsClassifier(n_neighbors=k,
p=2, # Euclidean distance
metric="minkowski")
classifier.fit(train_data, train_labels)
predictions = classifier.predict(test_data)
score = accuracy_score(test_labels, predictions)
X.append(k)
Y.append(score)
# Plot the accuracy vs. k
fig, ax = plt.subplots()
ax.set_xlabel('k (Number of Neighbors)')
ax.set_ylabel('Accuracy')
ax.plot(X, Y, "g-.o")
# ax.add_imageset_title('Accuracy vs Number of Neighbors')
ax.set_title( 'Accuracy vs. Number of Neighbors' )
plt.show()# Plot the accuracy vs. k
# fig, ax = plt.subplots()
# ax.set_xlabel('k (Number of Neighbors)')
# ax.set_ylabel('Accuracy')
# ax.plot(X, Y, "g-.o")
# ax.set_title('Accuracy vs Number of Neighbors')
# ax.set_title( 'Accuracy vs. Number of Neighbor')
# plt.show()
```
